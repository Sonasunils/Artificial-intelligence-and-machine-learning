```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create data using NumPy
np.random.seed(0)

students = ['A', 'B', 'C', 'D', 'E']
math = np.random.randint(60, 100, 5)
science = np.random.randint(50, 100, 5)
english = np.random.randint(55, 95, 5)
computer = np.random.randint(65, 100, 5)
history = np.random.randint(50, 90, 5)

# Step 2: Create Pandas DataFrame
df = pd.DataFrame({
    'Student': students,
    'Math': math,
    'Science': science,
    'English': english,
    'Computer': computer,
    'History': history
})

print("Student Data:\n")
print(df)

# Step 3: Set index
df.set_index('Student', inplace=True)

# Step 4: LINE CHART (Performance trend per student)
plt.figure()
for subject in df.columns:
    plt.plot(df.index, df[subject], marker='o', label=subject)

plt.title("Student Performance Line Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid()

plt.show()

# Step 5: BAR CHART (Average marks per subject)
avg_marks = df.mean()

plt.figure()
avg_marks.plot(kind='bar')

plt.title("Average Marks per Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)

plt.show()
```

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create data
np.random.seed(0)

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

laptop_sales = np.random.randint(20000, 50000, 6)
mobile_sales = np.random.randint(30000, 70000, 6)
tablet_sales = np.random.randint(15000, 40000, 6)

# Step 2: Create DataFrame
df = pd.DataFrame({
    'Month': months,
    'Laptop': laptop_sales,
    'Mobile': mobile_sales,
    'Tablet': tablet_sales
})

print("Sales Data:\n")
print(df)

# Step 3: Set index
df.set_index('Month', inplace=True)

# Step 4: LINE CHART (Sales trend)
plt.figure()

plt.plot(df.index, df['Laptop'], marker='o', label='Laptop')
plt.plot(df.index, df['Mobile'], marker='o', label='Mobile')
plt.plot(df.index, df['Tablet'], marker='o', label='Tablet')

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid()

plt.show()

# Step 5: BAR CHART (Total sales per product)
total_sales = df.sum()

plt.figure()
total_sales.plot(kind='bar')

plt.title("Total Sales per Product")
plt.xlabel("Products")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)

plt.show()
```

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Generate data
np.random.seed(2)

days = [f'Day{i}' for i in range(1, 11)]

day_temp = np.random.randint(25, 41, 10)    # 25°C to 40°C
night_temp = np.random.randint(18, 31, 10)  # 18°C to 30°C

# Step 2: Create DataFrame
df = pd.DataFrame({
    'Day': days,
    'Day Temp': day_temp,
    'Night Temp': night_temp
})

print("Weather Data:\n")
print(df)

# Step 3: Set index
df.set_index('Day', inplace=True)

# Step 4: LINE CHART (Day vs Night temperature)
plt.figure()

plt.plot(df.index, df['Day Temp'], marker='o', label='Day Temp')
plt.plot(df.index, df['Night Temp'], marker='o', label='Night Temp')

plt.title("Day vs Night Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid()

plt.show()


print(df)
# avg_temp = df.mean()
# print(avg_temp)
df['Avg Temp'] = (df['Day Temp'] + df['Night Temp']) / 2
print(df)

plt.figure()
plt.bar(df.index, df['Avg Temp'])

plt.title("Daily Average Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")

plt.show()
# Step 6: Analysis

# Hottest day
hottest_day = df['Day Temp'].idxmax()

# Coolest night
coolest_night = df['Night Temp'].idxmin()

print("\nHottest Day:", hottest_day)
print("Coolest Night:", coolest_night)

# Bonus: Temperature difference
df['Difference'] = df['Day Temp'] - df['Night Temp']

print("\nTemperature Difference:\n")
print(df['Difference'])

# Save to CSV
df.to_csv("weather_data.csv")
```
