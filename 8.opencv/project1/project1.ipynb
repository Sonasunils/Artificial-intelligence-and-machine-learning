```python
import cv2

# --------------------------------
# 1. Read Image
# --------------------------------

img = cv2.imread("flower.jpg")

if img is None:
    print("Image not found!")
    exit()

print("Image loaded successfully")


# --------------------------------
# 2. Display Image Dimensions
# --------------------------------

height, width, channels = img.shape

print("Original Width:", width)
print("Original Height:", height)
print("Number of Channels:", channels)


# --------------------------------
# 3. Display Original Image
# --------------------------------

cv2.imshow("Original Image", img)


# --------------------------------
# 4. Resize Image
# --------------------------------

resized = cv2.resize(img, (500, 300))

print("Resized Width: 500")
print("Resized Height: 300")

cv2.imshow("Resized Image", resized)


# --------------------------------
# 5. Convert Image to Grayscale
# --------------------------------

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale Image", gray)


# --------------------------------
# 6. Save Resized Image
# --------------------------------

cv2.imwrite("resized_image.png", resized)

print("Resized image saved successfully")


# --------------------------------
# 7. Save Grayscale Image
# --------------------------------

cv2.imwrite("gray_image.png", gray)

print("Grayscale image saved successfully")


# --------------------------------
# 8. Wait for Key
# --------------------------------

cv2.waitKey(0)


# --------------------------------
# 9. Close All Windows
# --------------------------------

cv2.destroyAllWindows()
```
