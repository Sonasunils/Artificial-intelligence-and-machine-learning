```python
import cv2
import numpy as np

# --------------------------------
# 1. Read Image
# --------------------------------

img = cv2.imread("fox.webp")

if img is None:
    print("Image not found!")
    exit()

print("Image loaded successfully")


# --------------------------------
# 2. Rotate Image
# --------------------------------

rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

cv2.imshow("Rotated Image", rotated)


# --------------------------------
# 3. Resize Image
# --------------------------------

resized = cv2.resize(img, (500, 300))

cv2.imshow("Resized Image", resized)


# --------------------------------
# 4. Flip Image
# --------------------------------

flipped = cv2
    .flip(img, 1)

cv2.imshow("Flipped Image", flipped)


# --------------------------------
# 5. Crop Image
# --------------------------------

cropped = img[100:400, 100:500]

cv2.imshow("Cropped Image", cropped)


# --------------------------------
# 6. Translate / Move Image
# --------------------------------

height, width = img.shape[:2]

matrix = np.float32([
    [1, 0, 100],
    [0, 1, 50]
])

translated = cv2.warpAffine(img, matrix, (width, height))

cv2.imshow("Translated Image", translated)


# --------------------------------
# 7. Save Images
# --------------------------------

cv2.imwrite("rotated.png", rotated)
cv2.imwrite("resized.png", resized)
cv2.imwrite("flipped.png", flipped)
cv2.imwrite("cropped.png", cropped)
cv2.imwrite("translated.png", translated)

print("All images saved successfully")


# --------------------------------
# 8. Wait and Close
# --------------------------------

cv2.waitKey(0)
cv2.destroyAllWindows()
```
