```python
import cv2

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Cannot open camera")
    exit()

# Read the first frame
ret, first_frame = cap.read()

if not ret:
    print("Cannot read camera")
    cap.release()
    exit()

# Convert first frame to grayscale
first_frame = cv2.cvtColor(
    first_frame,
    cv2.COLOR_BGR2GRAY
)

# Blur first frame
first_frame = cv2.GaussianBlur(
    first_frame,
    (21, 21),
    0
)

print("Camera started")
print("Press Q to exit")

while True:

    # Read current frame
    ret, frame = cap.read()

    if not ret:
        print("Cannot read frame")
        break

    # Convert current frame to grayscale
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # Blur
    gray = cv2.GaussianBlur(
        gray,
        (21, 21),
        0
    )

    # Compare first frame and current frame
    difference = cv2.absdiff(
        first_frame,
        gray
    )

    # Threshold
    _, threshold = cv2.threshold(
        difference,
        25,
        255,
        cv2.THRESH_BINARY
    )

    # Remove small gaps
    threshold = cv2.dilate(
        threshold,
        None,
        iterations=2
    )

    # Find contours
    contours, _ = cv2.findContours(
        threshold,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    motion = False

    # Check each contour
    for contour in contours:

        area = cv2.contourArea(contour)

        if area < 1000:
            continue

        motion = True

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Draw rectangle
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    # Display status
    if motion:

        cv2.putText(
            frame,
            "MOTION DETECTED",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )

    else:

        cv2.putText(
            frame,
            "NO MOTION",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (255, 0, 0),
            2
        )

    # ONLY ONE GUI WINDOW
    cv2.imshow(
        "Motion Detection",
        frame
    )

    # Press Q to exit
    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break

# Stop camera
cap.release()

# Close GUI
cv2.destroyAllWindows()

print("Camera stopped")
print("Program finished")
```
