```python
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
transactions = [
    ['Bread', 'Milk', 'Butter'],
    ['Bread', 'Milk'],
    ['Bread', 'Butter'],
    ['Milk', 'Butter'],
    ['Bread', 'Milk', 'Butter']
]
```

```python
te = TransactionEncoder()

te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(
    te_array,
    columns=te.columns_
)

print(df)
```

```python
frequent_items = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

print(frequent_items)
```

```python
rules = association_rules(
    frequent_items,
    metric='confidence',
    min_threshold=0.6
)

print(rules)
```

```python
result = rules[
    ['antecedents',
     'consequents',
     'support',
     'confidence',
     'lift']
]

print(result)
```
