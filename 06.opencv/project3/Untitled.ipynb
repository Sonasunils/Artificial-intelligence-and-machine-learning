```python
import cv2

# -----------------------------------------
# 1. Read Image
# -----------------------------------------

img = cv2.imread("images.jpg")

if img is None:
    print("Image not found!")
    exit()

print("Image loaded successfully")


# -----------------------------------------
# 2. Display Original Image
# -----------------------------------------

cv2.imshow("Original Image", img)


# -----------------------------------------
# 3. Convert to Grayscale
# -----------------------------------------

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale", gray)


# -----------------------------------------
# 4. Normal Blur
# -----------------------------------------

blur = cv2.blur(gray, (5, 5))

cv2.imshow("Normal Blur", blur)


# -----------------------------------------
# 5. Gaussian Blur
# -----------------------------------------

gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

cv2.imshow("Gaussian Blur", gaussian)


# -----------------------------------------
# 6. Median Blur
# -----------------------------------------

median = cv2.medianBlur(gray, 5)

cv2.imshow("Median Blur", median)


# -----------------------------------------
# 7. Thresholding
# -----------------------------------------

_, threshold = cv2.threshold(
    gray,
    127,
    255,
    cv2.THRESH_BINARY
)

cv2.imshow("Threshold", threshold)


# -----------------------------------------
# 8. Canny Edge Detection
# -----------------------------------------

canny = cv2.Canny(gray, 100, 200)

cv2.imshow("Canny Edge", canny)


# -----------------------------------------
# 9. Sobel Edge Detection
# -----------------------------------------

sobel_x = cv2.Sobel(
    gray,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)

sobel_y = cv2.Sobel(
    gray,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)

sobel = cv2.magnitude(sobel_x, sobel_y)

cv2.imshow("Sobel Edge", sobel)


# -----------------------------------------
# 10. Laplacian Edge Detection
# -----------------------------------------

laplacian = cv2.Laplacian(
    gray,
    cv2.CV_64F
)

cv2.imshow("Laplacian Edge", laplacian)


# -----------------------------------------
# 11. Pencil Sketch
# -----------------------------------------

inverted = cv2.bitwise_not(gray)

blurred = cv2.GaussianBlur(
    inverted,
    (21, 21),
    0
)

sketch = cv2.divide(
    gray,
    255 - blurred,
    scale=256
)

cv2.imshow("Pencil Sketch", sketch)


# -----------------------------------------
# 12. Save Results
# -----------------------------------------

cv2.imwrite("gray.png", gray)
cv2.imwrite("blur.png", blur)
cv2.imwrite("gaussian.png", gaussian)
cv2.imwrite("median.png", median)
cv2.imwrite("threshold.png", threshold)
cv2.imwrite("canny.png", canny)
cv2.imwrite("sobel.png", sobel)
cv2.imwrite("laplacian.png", laplacian)
cv2.imwrite("sketch.png", sketch)

print("All images saved successfully!")


# -----------------------------------------
# 13. Wait
# -----------------------------------------

cv2.waitKey(0)


# -----------------------------------------
# 14. Close Windows
# -----------------------------------------

cv2.destroyAllWindows()
```
