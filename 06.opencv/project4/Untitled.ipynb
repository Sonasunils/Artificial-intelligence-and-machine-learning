```python
import cv2
import math

# 1. Read image
img = cv2.imread("images.png")

if img is None:
    print("Image not found!")
    exit()

print("Image loaded successfully")


# 2. Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# 3. Threshold
_, threshold = cv2.threshold(
    gray,
    200,
    255,
    cv2.THRESH_BINARY_INV
)


# 4. Find contours
contours, hierarchy = cv2.findContours(
    threshold,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Total contours:", len(contours))


# 5. Check each contour
for contour in contours:

    # Area
    area = cv2.contourArea(contour)

    # Ignore small noise
    if area < 500:
        continue

    # Perimeter
    perimeter = cv2.arcLength(
        contour,
        True
    )

    # Approximate contour
    approx = cv2.approxPolyDP(
        contour,
        0.02 * perimeter,
        True
    )

    # Number of corners
    corners = len(approx)

    # Bounding rectangle
    x, y, w, h = cv2.boundingRect(contour)


    # -----------------------------------
    # 6. Shape Detection
    # -----------------------------------

    if corners == 3:

        shape = "Triangle"


    elif corners == 4:

        ratio = w / float(h)

        if 0.90 <= ratio <= 1.10:
            shape = "Square"
        else:
            shape = "Rectangle"


    elif corners > 6:

        circularity = (
            4 * math.pi * area
        ) / (perimeter * perimeter)

        if circularity > 0.80:
            shape = "Circle"
        else:
            continue


    else:

        # Ignore 5 or 6 corner shapes
        continue


    # -----------------------------------
    # 7. Draw contour
    # -----------------------------------

    cv2.drawContours(
        img,
        [contour],
        -1,
        (0, 255, 0),
        2
    )


    # -----------------------------------
    # 8. Write shape name
    # -----------------------------------

    cv2.putText(
        img,
        shape,
        (x, y - 5),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 0, 255),
        2
    )


    # -----------------------------------
    # 9. Print shape
    # -----------------------------------

    print(shape)


# -----------------------------------
# 10. Display result
# -----------------------------------

cv2.imshow(
    "Shape Detection",
    img
)


# -----------------------------------
# 11. Save result
# -----------------------------------

cv2.imwrite(
    "shape_detection_result.png",
    img
)
cv2.imshow("Threshold", threshold)

print("Result saved successfully!")


# -----------------------------------
# 12. Close
# -----------------------------------

cv2.waitKey(0)
cv2.destroyAllWindows()
```
