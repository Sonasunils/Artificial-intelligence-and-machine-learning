```python
import cv2

# ==========================================
# 1. Load Haar Cascade Face Detector
# ==========================================

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# Check whether Haar Cascade loaded correctly
if face_cascade.empty():
    print("Error: Face cascade could not be loaded")
    exit()

print("Face detector loaded successfully")


# ==========================================
# 2. Open Webcam
# ==========================================

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Cannot open webcam")
    exit()

print("Camera started")
print("Press Q to exit")


# ==========================================
# 3. Real-Time Face Detection
# ==========================================

while True:

    # Capture frame from webcam
    ret, frame = cap.read()

    if not ret:
        print("Error: Cannot read frame")
        break

    # ======================================
    # 4. Convert Frame to Grayscale
    # ======================================

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # ======================================
    # 5. Detect Faces
    # ======================================

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # ======================================
    # 6. Draw Rectangle Around Faces
    # ======================================

    for (x, y, w, h) in faces:

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Display FACE text
        cv2.putText(
            frame,
            "FACE",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )


    # ======================================
    # 7. Display Number of Faces
    # ======================================

    face_count = len(faces)

    cv2.putText(
        frame,
        "Faces: " + str(face_count),
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 0, 0),
        2
    )


    # ======================================
    # 8. Display Webcam
    # ======================================

    cv2.imshow(
        "Real-Time Face Detection",
        frame
    )


    # ======================================
    # 9. Press Q to Exit
    # ======================================

    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break


# ==========================================
# 10. Stop Camera
# ==========================================

cap.release()


# ==========================================
# 11. Close OpenCV Windows
# ==========================================

cv2.destroyAllWindows()

print("Camera stopped")
print("Program finished")
```
