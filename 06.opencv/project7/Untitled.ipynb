```python
import cv2
import os

# Name of the person
name = input("Enter person's name: ")

# Create folder
folder = os.path.join("dataset", name)

os.makedirs(folder, exist_ok=True)

# Load Haar Cascade
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

# Open webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Cannot open camera")
    exit()

count = 0

print("Look at the camera")
print("Press Q to quit")

while True:

    ret, frame = cap.read()

    if not ret:
        print("Cannot read frame")
        break

    # Convert to grayscale
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(100, 100)
    )

    for (x, y, w, h) in faces:

        # Draw rectangle
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Save face
        count += 1

        filename = os.path.join(
            folder,
            f"{count}.jpg"
        )

        face_image = gray[
            y:y + h,
            x:x + w
        ]

        cv2.imwrite(
            filename,
            face_image
        )

        cv2.putText(
            frame,
            f"Images: {count}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

    cv2.imshow(
        "Collect Faces",
        frame
    )

    # Stop after 30 images
    if count >= 30:
        break

    # Press Q
    if cv2.waitKey(100) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()

print("Face collection completed")
print("Images saved:", count)
```
